```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
import numpy as np

df = pd.read_excel('/content/drive/MyDrive/ABC Company.xlsx')

print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
np.random.seed(42)
df['Height'] = np.random.randint(150,181,size= len(df))
df.head()
```

```python
df.tail()
```

```python
emp_team = df['Team'].value_counts()
print(emp_team)
```

```python
emp_team_percent =(emp_team / len(df)) *100
print(emp_team_percent)
```

```python
emp_position = df['Position'].value_counts()
print(emp_position)
```

```python
age_group = df['Age'].mode()[0]
print(age_group)
```

```python
salary_grp = df.groupby(['Team','Position'])['Salary'].sum()

print(salary_grp)
highest_salary =salary_grp.sort_values( ascending= False)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
cr_matrix =df[['Age', 'Salary']].corr()
sns.heatmap(cr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix: Age and Salary')
plt.show()
```

```python
#Graphical Representation
plt.pie(emp_team_percent, labels=emp_team.index, autopct='%1.1f%%')
plt.title('Employee Team Distribution')
plt.show()
```

```python
x = emp_position.index
y = emp_position.values
plt.bar(x,y)
plt.title('Position VS Number of Employees')
plt.xlabel('Position')
plt.ylabel('Number of Employees')
plt.show()


```

```python
plt.hist(df['Age'], bins =10, edgecolor ='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('No of Employees')
plt.show()
```
